**Remove Outliers**

```python
Q1=df['Price'].quantile(0.25)
Q3=df['Price'].quantile(0.75)
IQR=Q3-Q1
df = df[(df['Price'] >= Q1-1.5*IQR) & (df['Price'] <= Q3+1.5*IQR)]
```

# **Key Questions**


```python
#1. Do men tend to prefer SUVs more compared to women?
sns.countplot(x='Gender',hue='Make',data=df)
plt.title("Gender vs Car Type")
plt.show()
```

Female customers have a higher number of SUV purchases compared to male customers.

Male customers mainly purchase Hatchbacks and Sedans.

Women shows a stronger preference for SUVs than men.

```python
#2. What is the likelihood of a salaried person buying a Sedan?
sns.countplot(x='Profession',hue='Make',data=df)
plt.title("Car Type preference by Profession")
plt.show()
```

Among salaried customers, Sedans appear to be the most commonly purchased automobile type.

A significant proportion of salaried individuals choose Sedans over SUVs and Hatchbacks.

Sedans may be preferred by salaried individuals because they provide a balance between affordability, comfort, and practicality.


```python
#3. What evidence or data supports Sheldon Cooper's claim that a salaried male is an easier target for a SUV sale over a Sedan sale?
salaried_males=df[(df['Profession']=='Salaried') & (df['Gender']=='Male')]
sns.countplot(x='Make',data=salaried_males)
plt.title("Car Type preference by salaried males")
plt.show()
```

 Among salaried males, Sedan purchases appear to be higher than SUV purchases.

 Sheldon Cooper’s claim is not supported

```python
#4. How does the the amount spent on purchasing automobiles vary by gender?
sns.barplot(x='Gender',y='Price',data=df)
plt.title("Average spending by Gender")
plt.show()
```

Female customers tend to spend slightly more on automobiles compared to male customers.

This may be because female buyers in the dataset appear to prefer SUVs, which generally fall in the higher price range.

```python
#5. How much money was spent on purchasing automobiles by individuals who took a personal loan?
sns.barplot(x='Personal_loan',y='Price',data=df)
plt.title("Average spending by personal loan status")
plt.show()
```

Customers with personal loans still spend a considerable amount on automobiles.
The average spending of customers without personal loans appears slightly higher.

```python
#6.Working partner influence
sns.boxplot(x='Partner_working',y='Price',data=df)
plt.title("Average spending by partner working status")
plt.show()
```

Customers with working partners tend to have higher total household income.

This enabling them to consider higher-priced automobiles.

**Actionable Insights**

1.	Female customers show a stronger preference for SUVs compared to male customers. This suggests that gender plays a role in automobile type preference.
2.	Sedans are the most commonly purchased vehicles among salaried professionals, indicating that this segment values practicality, affordability, and comfort.
3.	Income level has a positive relationship with automobile price, meaning customers with higher salaries or higher household income tend to purchase more expensive vehicles.

**Business Recommendations**

1.	Target Female Customers for SUV Marketing
Automobile companies can design marketing campaigns that highlight SUV features such as safety, comfort, and space, which may appeal to female customers.
